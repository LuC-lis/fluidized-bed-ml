单个数据pvd to excel

```python
import os
import pandas as pd

# 文件夹路径
folder_path = r"F:\XC_OFP&2D camera\FCC110\80cm"
# 输出文件夹路径
output_folder_path = r"C:\Users\cbi20\Desktop\OFP data XC\output"
os.makedirs(output_folder_path, exist_ok=True)  # 创建输出文件夹，如果不存在
output_file = os.path.join(output_folder_path, "80cm.xlsx")  # 初始化一个空的DataFrame用于存放所有数据
all_data = pd.DataFrame()

# 遍历文件夹中的所有pvd文件
for filename in os.listdir(folder_path):
    if filename.endswith(".pvd"):
        file_path = os.path.join(folder_path, filename)

        # 读取文件并跳过前15行
        with open(file_path, 'r') as file:
            lines = file.readlines()[15:]

        # 初始化临时列表用于存储一列数据
        ch2_values = []

        # 处理每一行数据
        for line in lines:
            # 去掉首尾空白字符
            line = line.strip()
            # 按两个空格分割，确保分割成两部分
            parts = line.split('  ')

            # 确保分割后有两部分数据
            if len(parts) >= 2:
                try:
                    ch2_values.append(float(parts[1]))

                except ValueError:
                    print(f"无法转换为float类型的值: {parts} 在文件 {filename} 中")
            else:
                print(f"行格式不正确（未找到两个空格分隔符）：'{line}' 在文件 {filename} 中")

        # 将一列数据放入一个DataFrame，并在第一行设置为文件名
        temp_df = pd.DataFrame({
            filename: ch2_values
        })
        all_data = pd.concat([all_data, temp_df], axis=1)

# 将结果写入到Excel中
all_data.to_excel(output_file, index=False)

print(f"数据已成功处理并保存到 {output_file}")
```

提取平均值

```python
import os
import pandas as pd

# 定义主文件夹路径
main_folder = "output results"  # 请替换为你的主文件夹路径

# 定义所有子文件夹名称
sub_folders = ["h40", "h60", "h80", "h102"]
sub_sub_folders = ["r0", "r0.316", "r0.548","r0.707","r0.837","r0.95"]

# 创建一个空的DataFrame用于汇总结果
summary_df = pd.DataFrame()

# 遍历每个主文件夹下的子文件夹
for sub_folder in sub_folders:
    for sub_sub_folder in sub_sub_folders:
        folder_path = os.path.join(main_folder, sub_folder, sub_sub_folder)

        # 检查路径是否存在
        if not os.path.exists(folder_path):
            print(f"路径 {folder_path} 不存在，跳过...")
            continue

        # 遍历文件夹中的所有Excel文件
        for file_name in os.listdir(folder_path):
            if file_name.endswith(".xlsx"):  # 确保是Excel文件
                file_path = os.path.join(folder_path, file_name)

                # 读取Excel文件的最后一行数据
                try:
                    df = pd.read_excel(file_path)
                    last_row = df.iloc[-1].copy()  # 提取最后一行
                    last_row["文件名"] = file_name  # 添加文件名列
                    last_row["子文件夹"] = sub_sub_folder  # 添加子文件夹名称
                    last_row["主文件夹"] = sub_folder  # 添加主文件夹名称
                    summary_df = pd.concat([summary_df, last_row.to_frame().T], ignore_index=True)
                except Exception as e:
                    print(f"处理文件 {file_path} 时出错: {e}")

# 将汇总结果保存到一个新的Excel文件中
summary_file_path = os.path.join(main_folder, "汇总结果.xlsx")
summary_df.to_excel(summary_file_path, index=False)
print(f"汇总结果已保存到 {summary_file_path}")
```
