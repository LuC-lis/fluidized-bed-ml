```python
import os
import pandas as pd
import numpy as np

# 文件夹路径
folder_path = (r"C:\Users\cbi20\Desktop\OFP data XC\output\h102\r0.95")
output_folder_path = r"C:\Users\cbi20\Desktop\OFP data XC\output results\h102\r0.95"
os.makedirs(output_folder_path, exist_ok=True)  # 创建输出文件夹，如果不存在

# 获取文件夹中的所有文件名
file_names = [f for f in os.listdir(folder_path) if f.endswith('.xlsx')]

# 遍历每个文件
for file_name in file_names:
    # 读取数据
    file_path = os.path.join(folder_path, file_name)
    df = pd.read_excel(file_path)

    # 获取数据的行和列数
    rows, cols = df.shape

    # 初始化一个 DataFrame 来存储结果
    results = pd.DataFrame(index=range(rows + 1))  # +1 用于存储平均值

    # 遍历每一列
    for j in range(cols):
        # 取出一列数据
        col_data = df.iloc[:, j]

        # 应用公式进行计算
        calculated_data = 0.0164 * np.exp(0.7759 * col_data / 1000)

        # 将原始数据和计算结果存入结果 DataFrame
        results[f'Original_{j+1}'] = col_data #存储原始数据
        results[f'Calculated_{j+1}'] = calculated_data#存储计算数据

        # 计算每列的平均值，并将结果存入结果 DataFrame
        mean_value = calculated_data.mean()
        results.loc[rows, f'Calculated_{j+1}'] = mean_value

    # 提取最后三列的平均值形成一个 Series
    mean_series = results.iloc[rows, -3:]  # 提取最后三列

    # 对这个 Series 求平均值
    final_mean = mean_series.mean()

    # 生成输出文件名，并将最终的平均值显示在文件名中
    base_name = os.path.splitext(file_name)[0]
    output_file_name = f'OFP-Results-{base_name}-final_mean_{final_mean:.4f}.xlsx'  # 保留4位小数
    output_file_path = os.path.join(output_folder_path, output_file_name)

    # 写入结果到新的 Excel 文件
    results.to_excel(output_file_path, index=False)

    print(f"计算结果已保存到 {output_file_path}")
```
